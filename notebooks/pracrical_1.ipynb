## 5. Распределение длины моллюсков

```python
plt.figure(figsize=(8, 5))

sns.histplot(df["Length"], kde=True)

plt.title("Распределение длины моллюсков")
plt.xlabel("Length")
plt.ylabel("Количество")

plt.show()
```

## 6. Соотношение категорий по полу и возрастной группе

```python
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x="gender")

plt.title("Соотношение категорий gender")
plt.xlabel("Категория")
plt.ylabel("Количество")

plt.show()
```

```python
df["gender"].value_counts()
```

```python
df.groupby("gender")[["Length", "Whole weight", "Rings"]].mean()
```

## 7. Наблюдения

1. Распределение длины моллюсков показывает, что большинство наблюдений сосредоточено примерно в диапазоне от 0.4 до 0.7, тогда как очень маленькие и очень большие значения встречаются значительно реже.

2. Категории M, F и I представлены в датасете неравномерно: категория M встречается чаще, чем F и I, однако все три категории имеют достаточно большое количество наблюдений.

3. Категория I в среднем имеет меньшие значения длины и массы по сравнению с категориями M и F, что соответствует различиям между молодыми и взрослыми особями.
